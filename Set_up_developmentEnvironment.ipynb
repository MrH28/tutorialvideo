# VS Code Development Environment Workflow

Use this workflow to set up, check, and maintain a clean Python environment in VS Code.

1. Confirm Python, Git, and the VS Code Python extension.
2. Open the project folder.
3. Create and activate `.venv`.
4. Select the interpreter or notebook kernel.
5. Install the required packages.
6. Verify the setup.
7. Work, test, and commit in small steps.

1. Open the project folder in VS Code.

4. Activate it.

## How to Create & Activate a `.venv` (step-by-step) ?

1) Create the virtual environment in your project root:
```bash
python -m venv .venv
```

2) Activate the environment (choose the command for your shell):
- PowerShell (Windows):
```powershell
.\.venv\Scripts\Activate.ps1
```
- Command Prompt (Windows):
```cmd
.venv\Scripts\activate
```
- Git Bash / MinGW (Windows):
```bash
source .venv/Scripts/activate
```
- macOS / Linux:
```bash
source .venv/bin/activate
```

3) Upgrade pip and wheel (recommended):
```bash
python -m pip install --upgrade pip setuptools wheel
```

4) Verify activation:
```python
import sys
print(sys.executable)
```
The printed path should point inside `.venv` (e.g. `.venv\Scripts\python.exe`).

5) Use the environment in VS Code: select the interpreter that points to `.venv` (status bar or `Python: Select Interpreter`).

Notes / Troubleshooting:
- If PowerShell blocks activation, you can allow scripts for the current user:
```powershell
Set-ExecutionPolicy -ExecutionPolicy RemoteSigned -Scope CurrentUser
```
- If your system uses an externally-managed Python (PEP 668), prefer creating and activating a project `.venv` instead of installing into the system Python.
- To make the environment available as a Jupyter kernel (optional):
```bash
pip install ipykernel
python -m ipykernel install --user --name=project-venv --display-name "Python (project-venv)"
```

## Select Interpreter / Notebook Kernel & Install Packages

### A) Selecting the interpreter in VS Code
- Click the Python interpreter shown in the status bar (bottom-right) and pick the one located inside `.venv` (path will include `.venv`).
- Or open Command Palette: `Ctrl+Shift+P` → `Python: Select Interpreter` → choose `.venv`.
- For notebooks: open the `.ipynb`, click the kernel name at the top-right, and select the kernel matching your `.venv` interpreter.
- Verify from a cell:
```python
import sys
print(sys.executable)
```
The printed path should point into `.venv` (e.g. `.venv\Scripts\python.exe`).

### B) Recommended packages to install (starter set)
Run this once the `.venv` is active and selected in VS Code:\
```bash
pip install jupyter ipykernel numpy pandas matplotlib seaborn scikit-learn
```
Short purpose of each package:
- `jupyter`: run and manage notebooks (server).
- `ipykernel`: register your virtualenv as a notebook kernel.
- `numpy`: efficient numeric arrays and math primitives.
- `pandas`: data frames, CSV I/O, data cleaning utilities.
- `matplotlib`: basic plotting library for figures and screenshots.
- `seaborn`: statistical visualization built on matplotlib (nicer defaults).
- `scikit-learn`: common machine learning algorithms and helpers.

### C) Save & reproduce the environment
- Save installed packages to `requirements.txt`:
```bash
pip freeze > requirements.txt
```
- Recreate on another machine:
```bash
pip install -r requirements.txt
```

### D) Make the `.venv` available as a Jupyter kernel (optional but recommended)
```bash
pip install ipykernel
python -m ipykernel install --user --name=project-venv --display-name "Python (project-venv)"
```
After this, the kernel list in notebooks will show `Python (project-venv)`.

### E) Quick verification checklist
- Confirm `sys.executable` points to `.venv`.
- Run `import pandas as pd; import numpy as np; import matplotlib` to ensure imports succeed.
- Open a notebook, select the new kernel, and run a plot cell to confirm display works.



## Verify Setup, Use Git, and Daily Workflow

### 1) Quick verification steps (run in selected kernel)
- Confirm interpreter path and Python version:
```python
import sys
print(sys.executable)
print(sys.version)
```
- Check imports used by your project:
```python
import numpy as np
import pandas as pd
import matplotlib
print(np.__version__, pd.__version__)
```
- Small end-to-end test: read a tiny CSV and plot one column:
```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.DataFrame({'x':[1,2,3],'y':[2,4,3]})
display(df.head())
df.plot(x='x', y='y', kind='line', title='Smoke test')
plt.show()
```
If all of these run without errors, your environment and kernel are correctly configured.

### 2) Using Git with the environment
- Create `.gitignore` (exclude `.venv` and caches):
```text
.venv/
__pycache__/
*.pyc
.ipynb_checkpoints/
```
- Basic repository workflow (one-time init):
```bash
git init
git add .
git commit -m "Initial commit - project scaffold"
git branch -M main
git remote add origin <your-repo-url>
git push -u origin main
```
- Daily Git habits:
  - Pull latest changes before you start: `git pull`.
  - Commit small logical changes with clear messages.
  - Add `requirements.txt` after package changes: `pip freeze > requirements.txt` and commit it.

### 3) Daily workflow (copyable checklist)
1. Open VS Code and the project folder.
2. Activate `.venv` and select the interpreter/kernel.
3. Pull latest changes: `git pull`.
4. Run a quick smoke test (one notebook cell or script).
5. Implement one focused task or fix.
6. Run tests or verify functionality locally.
7. Update `requirements.txt` if you added packages: `pip freeze > requirements.txt`.
8. Stage and commit: `git add -A && git commit -m "Short: what I changed"`.
9. Push your branch or main: `git push`.

### 4) Troubleshooting tips
- If imports fail, ensure the kernel's `sys.executable` points to `.venv` and re-install packages into that environment.
- If display/plots don't show in notebooks, confirm the notebook uses the correct kernel and try `%matplotlib inline` or `%matplotlib widget`.
- For problems installing packages in managed system Python, create & use a project `.venv` and install there.
