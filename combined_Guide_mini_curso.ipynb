## 4 — Boxplot example (visualization)

The following cell (from the original guide) computes and draws a grouped boxplot using an SVG renderer. It expects a CSV with numeric values and a grouping column — adjust `file_path`, `value_column` and `group_column`.

````python
# Adapted boxplot SVG code from Guide.ipynb — set your file path first
import csv
from collections import defaultdict
from html import escape
from statistics import quantiles
from IPython.display import SVG, display

# Update this path to your CSV if you want to run the SVG renderer
file_path = r"c:\Users\PC\Desktop\Mini_curso\BSSL_SoilAttributes_SP.csv"
value_column = "SOM_gkg"
group_column = "Geology"

values_by_group = defaultdict(list)
try:
    with open(file_path, 'r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        for row in reader:
            group_value = row.get(group_column, '').strip()
            raw_value = row.get(value_column, '').strip()
            if group_value and raw_value:
                try:
                    values_by_group[group_value].append(float(raw_value))
                except ValueError:
                    pass
except FileNotFoundError:
    display(SVG('<svg width=
600
 height=
40
><text x=
10
 y=
20
>CSV file not found — update file_path</text></svg>'))

groups = [g for g, v in values_by_group.items() if v]
if groups:
    box_stats = []
    all_values = []
    for group in groups:
        data = sorted(values_by_group[group])
        all_values.extend(data)
        q1, q2, q3 = quantiles(data, n=4, method='inclusive')
        iqr = q3 - q1
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr
        whisker_low = min(value for value in data if value >= lower_bound)
        whisker_high = max(value for value in data if value <= upper_bound)
        outliers = [value for value in data if value < lower_bound or value > upper_bound]
        box_stats.append({
            'group': group,
            'q1': q1, 'median': q2, 'q3': q3,
            'whisker_low': whisker_low, 'whisker_high': whisker_high, 'outliers': outliers
        })
    # Simple SVG output to display a placeholder summary (detailed SVG omitted here)
    display(SVG('<svg width=
600
 height=
40
><text x=
10
 y=
20
>Boxplot computed for groups: ' + ', '.join(groups[:6]) + '</text></svg>'))
else:
    display(SVG('<svg width=
600
 height=
40
><text x=
10
 y=
20
>No groups/data found in CSV</text></svg>'))
````
