## How to use Copilot Chat to plan and build a script

Use Copilot Chat like a partner for writing code in two steps:

1. Give it the **context**.
2. Ask for the **plan**.
3. Ask for the **script**.

### How to structure the context

Before you ask for code, include:

- **Goal**: what you want to build.
- **Input**: files, data, or notebook cells involved.
- **Output**: what the final result should look like.
- **Constraints**: style, language, libraries, or limits.
- **Current state**: what already works and what is broken.

### Example prompt for planning

> I have a notebook that analyzes soil data. The input is a CSV file, the output should be a clean report and a simple model, and I want to use Python and pandas. Please read the context and give me a step-by-step plan before writing any code.

### Example prompt for the script

> Now use that plan and write the Python script step by step. Keep the code simple, explain each part briefly, and make it ready to run in a notebook cell.

### Good chat workflow

- Start with a short summary of the task.
- Paste the relevant cell, error, or file names.
- Ask for a plan first if the task is large.
- Ask for the script after you approve the plan.
- Review the code before running it.

### Example of a full request

> I want to build a small notebook that loads a CSV, cleans missing values, trains a model, and prints the result. The data is in `soil_data.csv`. Please first give me a plan, then write the script.

### Simple rule

If the task is unclear, ask for a plan. If the task is clear, ask for the script. If the task is large, do both in order.

## GitHub commands: commit, pull, push, and branches

### Basic idea

GitHub is where your code is stored online. Git is the tool you use on your computer to track changes.

### What each command does

- `git commit`: saves a snapshot of your work locally on your computer.
- `git pull`: downloads the newest changes from GitHub into your local project.
- `git push`: sends your local commits up to GitHub.
- `git branch`: shows or creates branches, which are separate lines of work.

### Simple example

Imagine you changed a file in your project.

1. You edit the file.
2. You run `git add .` to stage the changes.
3. You run `git commit -m "Fix notebook explanation"` to save the snapshot.
4. You run `git push` to send that snapshot to GitHub.

If someone else changed the same project on GitHub, you run `git pull` first to get their updates before continuing.

### What is a branch?

A branch is like a separate copy of the project for a specific task.

Example:

- `main` branch: the stable version.
- `feature/docs` branch: used for adding or testing new text.
- `bugfix/data-cleaning` branch: used for fixing a problem.

### Very short workflow

- Create a branch for new work.
- Make changes.
- Commit the changes.
- Push the branch to GitHub.
- Open a pull request if you want to merge it into `main`.

### Useful commands

```bash
git status
git branch
git checkout -b feature/docs
git add .
git commit -m "Add GitHub explanation"
git push
```

### Easy rule

- `commit` = save locally
- `push` = send to GitHub
- `pull` = get from GitHub
- `branch` = separate line of work

## What is a boxplot and how to use it

A **boxplot** is a chart that shows how data is spread out. It helps you see:

- the median, which is the middle value
- the first and third quartiles, which show the middle 50% of the data
- possible outliers, which are unusual values
- whether the data is skewed to one side

### Why use it

Use a boxplot when you want to compare groups or check if your data has unusual values.

### Simple example in Python

```python
import pandas as pd
import matplotlib.pyplot as plt

# Example data
df = pd.DataFrame({
    "group": ["A", "A", "A", "B", "B", "B"],
    "value": [10, 12, 14, 20, 22, 40]
})

# Boxplot by group
df.boxplot(column="value", by="group")
plt.title("Boxplot by Group")
plt.suptitle("")
plt.xlabel("Group")
plt.ylabel("Value")
plt.show()
```

### How to read it

- The line inside the box is the median.
- The box shows the middle 50% of the data.
- The whiskers show the spread beyond the box.
- Dots outside the whiskers are possible outliers.

### Easy rule

- Use a boxplot to compare distributions.
- Use a histogram to see the overall shape of the data.

> If you meant a different chart name than boxplot, I can rewrite this section.

````python
import csv
from collections import defaultdict
from html import escape
from statistics import quantiles
from IPython.display import SVG, display

# Load your soil dataset
file_path = r"c:\Users\PC\Desktop\Mini_curso\BSSL_SoilAttributes_SP.csv"

# Pick one soil attribute and one grouping column
value_column = "SOM_gkg"
group_column = "Geology"

# Read values by group using the standard library
values_by_group = defaultdict(list)

with open(file_path, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        group_value = row.get(group_column, "").strip()
        raw_value = row.get(value_column, "").strip()
        if group_value and raw_value:
            try:
                values_by_group[group_value].append(float(raw_value))
            except ValueError:
                pass

# Keep only groups that have data
groups = [group for group, values in values_by_group.items() if values]
if not groups:
    raise ValueError(f"No data found for {group_column} and {value_column}")

# Compute boxplot stats for each group
box_stats = []
all_values = []
for group in groups:
    data = sorted(values_by_group[group])
    all_values.extend(data)
    q1, q2, q3 = quantiles(data, n=4, method="inclusive")
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    whisker_low = min(value for value in data if value >= lower_bound)
    whisker_high = max(value for value in data if value <= upper_bound)
    outliers = [value for value in data if value < lower_bound or value > upper_bound]
    box_stats.append({
        "group": group,
        "q1": q1,
        "median": q2,
        "q3": q3,
        "whisker_low": whisker_low,
        "whisker_high": whisker_high,
        "outliers": outliers,
    })

# SVG layout
width = 1200
height = 650
margin_left = 90
margin_right = 40
margin_top = 70
margin_bottom = 120
plot_width = width - margin_left - margin_right
plot_height = height - margin_top - margin_bottom
min_value = min(all_values)
max_value = max(all_values)
if max_value == min_value:
    max_value += 1


def y_scale(value):
    return margin_top + plot_height - ((value - min_value) / (max_value - min_value)) * plot_height

# Draw the SVG figure
box_width = max(20, min(70, plot_width / max(1, len(box_stats) * 2)))
step = plot_width / max(1, len(box_stats))
svg_parts = [
    f'<svg xmlns="http://www.w3.org/2000/svg" width="{width}" height="{height}" viewBox="0 0 {width} {height}">',
    '<rect width="100%" height="100%" fill="white"/>',
    f'<text x="{width / 2}" y="35" text-anchor="middle" font-size="22" font-family="Arial">Boxplot of {escape(value_column)} by {escape(group_column)}</text>',
    f'<line x1="{margin_left}" y1="{margin_top}" x2="{margin_left}" y2="{margin_top + plot_height}" stroke="black"/>',
    f'<line x1="{margin_left}" y1="{margin_top + plot_height}" x2="{width - margin_right}" y2="{margin_top + plot_height}" stroke="black"/>',
]

# y-axis labels
for tick in range(5):
    value = min_value + (max_value - min_value) * tick / 4
    y = y_scale(value)
    svg_parts.append(f'<line x1="{margin_left - 5}" y1="{y}" x2="{margin_left}" y2="{y}" stroke="black"/>')
    svg_parts.append(
        f'<text x="{margin_left - 10}" y="{y + 4}" text-anchor="end" font-size="11" font-family="Arial">{value:.1f}</text>'
    )

for index, stat in enumerate(box_stats):
    x_center = margin_left + step * (index + 0.5)
    q1_y = y_scale(stat["q1"])
    median_y = y_scale(stat["median"])
    q3_y = y_scale(stat["q3"])
    whisker_low_y = y_scale(stat["whisker_low"])
    whisker_high_y = y_scale(stat["whisker_high"])

    # Whiskers
    svg_parts.append(f'<line x1="{x_center}" y1="{whisker_high_y}" x2="{x_center}" y2="{q3_y}" stroke="#333" stroke-width="2"/>')
    svg_parts.append(f'<line x1="{x_center}" y1="{q1_y}" x2="{x_center}" y2="{whisker_low_y}" stroke="#333" stroke-width="2"/>')
    svg_parts.append(f'<line x1="{x_center - box_width / 3}" y1="{whisker_high_y}" x2="{x_center + box_width / 3}" y2="{whisker_high_y}" stroke="#333" stroke-width="2"/>')
    svg_parts.append(f'<line x1="{x_center - box_width / 3}" y1="{whisker_low_y}" x2="{x_center + box_width / 3}" y2="{whisker_low_y}" stroke="#333" stroke-width="2"/>')

    # Box
    svg_parts.append(
        f'<rect x="{x_center - box_width / 2}" y="{q3_y}" width="{box_width}" height="{q1_y - q3_y}" fill="#cfe8ff" stroke="#1f4e79" stroke-width="2"/>'
    )
    # Median
    svg_parts.append(f'<line x1="{x_center - box_width / 2}" y1="{median_y}" x2="{x_center + box_width / 2}" y2="{median_y}" stroke="#1f4e79" stroke-width="2"/>')

    # Outliers
    for outlier in stat["outliers"]:
        y = y_scale(outlier)
        svg_parts.append(f'<circle cx="{x_center}" cy="{y}" r="3" fill="#d62728"/>')

    # Group label
    svg_parts.append(
        f'<text x="{x_center}" y="{height - 70}" text-anchor="middle" font-size="11" font-family="Arial">{escape(stat["group"][:22])}</text>'
    )

svg_parts.append(
    f'<text x="{width / 2}" y="{height - 35}" text-anchor="middle" font-size="13" font-family="Arial">{escape(group_column)}</text>'
)
svg_parts.append(
    f'<text x="25" y="{margin_top + plot_height / 2}" text-anchor="middle" font-size="13" font-family="Arial" transform="rotate(-90 25 {margin_top + plot_height / 2})">{escape(value_column)}</text>'
)
svg_parts.append('</svg>')

# Display the full figure in the notebook
svg = "\n".join(svg_parts)
display(SVG(svg))
````

## Interpreting the boxplot of SOM_gkg by Geology

This boxplot shows that **SOM_gkg changes by geology type**.

- **Igneous Rock** has the **highest median** SOM_gkg, around **16.5**, so soils in this group tend to have more organic matter.
- **Metamorphic Rock** is in the middle, with a median around **14.0**.
- **Sedimentary Rock** has the **lowest median**, around **12.6**, so it tends to show lower SOM_gkg values overall.

The spread is also different across groups:

- **Igneous Rock** and **Sedimentary Rock** have many samples, so their boxplots are more reliable for comparison.
- **Metamorphic Rock** has a much smaller sample size, so its pattern should be interpreted more carefully.
- **Sedimentary Rock** shows the **largest number of outliers**, which suggests more variability in SOM_gkg.

### Simple conclusion

The geological group appears to influence SOM_gkg. In this dataset, **Igneous Rock tends to have higher SOM_gkg than Sedimentary Rock**, while **Metamorphic Rock sits between them**.
